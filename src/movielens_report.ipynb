## Результаты методов класса Movies

```python
cls_movies = MA.Movies('../datasets/movies.csv')
%timeit MA.Movies('../datasets/movies.csv')
```

### Метод dist_by_release() возвращает словарь, упорядоченный по убыванию, в котором ключи - год выхода, значения - количество фильмов, выпущенных в этом году.

```python
films_by_year = cls_movies.dist_by_release()
print(films_by_year)
%timeit cls_movies.dist_by_release()
```

### Метод dist_by_genres() возвращает словарь, упорядоченный по убыванию, в котором ключи - жанр, значения - количество фильмов по этому жанру.

```python
films_by_genres = cls_movies.dist_by_genres()
print(films_by_genres)
%timeit cls_movies.dist_by_genres()
```

### Метод most_genres(n) возвращает словарь с топ-n позициями, упорядоченный по убыванию, в котором ключи - название фильма, значения - количество жанров в этом фильме.

```python
films_by_genres = cls_movies.most_genres(20)
print(films_by_genres)
%timeit cls_movies.most_genres(20)
```

## Результаты методов класса Tags

```python
cls_tags = MA.Tags('../datasets/tags.csv')
%timeit MA.Tags('../datasets/tags.csv')
```

### Метод most_words(n) возвращает словарь,упорядоченный по убыванию, состоящий из топ-n тэгов, в которых содержится больше всего слов. Ключи - название тэга, значения - кол-во слов в тэге

```python
tags_most_words = cls_tags.most_words(20)
print(tags_most_words)
%timeit cls_tags.most_words(20)
```

### Метод longest(n) возвращает список, упорядоченный по убыванию, состоящий из топ-n длинных тэгов по кол-ву символов

```python
tags_longest = cls_tags.longest(20)
print(tags_longest)
%timeit cls_tags.longest(20)
```

### Метод most_words_and_longest(n) возвращает неупорядоченный список пересечений между топ-n самых длинных тэгов по словам и по топ-n самых длинных по символам

```python
tags_all_long = cls_tags.most_words_and_longest(20)
print(tags_all_long)
%timeit cls_tags.most_words_and_longest(20)
```

### Метод most_popular(n) возвращает словарь с топ-n тэгами по кол-ву использования, упорядоченный по убыванию, где ключи - название тэга, значения - кол-во использований

```python
popular_tags = cls_tags.most_popular(20)
print(popular_tags)
%timeit cls_tags.most_popular(20)
```

### Метод tags_with(word) возвращает сортированный по алфавиту список тэгов, содержащие слово word

```python
tags_with_word = cls_tags.tags_with('romance')
print(tags_with_word)
%timeit cls_tags.tags_with('romance')
```

```python
tags_with_word = cls_tags.tags_with('blabla')
print(tags_with_word)
%timeit cls_tags.tags_with('blabla')
```

## Результаты методов класса Links

```python
filepath = '../datasets/links.csv'
cls_links = MA.Links(filepath)
%timeit MA.Links(filepath)
```

### Метод get_imdb(list_of_movies, list_of,fields) возвращает сортированный по убыванию MovieID список списков формата [movieID, field1, field2 ...] для списка фильмов, переданных в качестве аргумента. fields - поля, которые парсятся с сайта IMDB

```python
res = cls_links.get_imdb([181065, 1, 5,200000,6666666666], ['movieID', 'Title','Genre', 'Director', 'Cumulative Worldwide Gross','blablabla'])
print(*res, sep='\n')
%timeit cls_links.get_imdb([181065, 1, 5,200000,6666666666], ['movieID', 'Title','Genre', 'Director', 'Cumulative Worldwide Gross','blablabla'])
```

### Метод top_directors(n) возвращает словарь с топ-n позициями, где ключи - режиссеры, значения - кол-во снятых фильмов. Сортирован по убыванию по кол-ву фильмов у режиссера

```python
top_dir = cls_links.top_directors(20)
print(top_dir)
%timeit cls_links.top_directors(20)
```

### Метод most_expensive(n) возвращает словарь с топ-n позициями, где ключи - названия фильмов, значения - бюджет фильма. Сортирован по убыванию по бюджету фильма.

```python
most_exp = cls_links.most_expensive(20)
print(most_exp)
%timeit cls_links.most_expensive(20)
```

### Метод most_profitable(n) возвращает словарь с топ-n позициями, где ключи - названия фильмов, значения - разница между сборами и бюджетом. Сортирован по убыванию по прибыльности фильма.

```python
most_prof = cls_links.most_profitable(20)
print(most_prof)
%timeit cls_links.most_profitable(20)
```

### Метод longest(n) возвращает словарь с топ-n позициями, где ключи - названия фильмов, значения - длительность фильма. Сортирован по убыванию по длительности фильма. Если версий больше одной - выбираем любую.

```python
most_long = cls_links.longest(20)
print(most_long)
%timeit cls_links.longest(20)
```

### Метод top_cost_per_minute(n) возвращает словарь с топ-n позициями, где ключи - названия фильмов, значения - бюджет фильма/ длительность фильма. Сортирован по убыванию по показателю. Не требуется приводить бюджет к одной валюте, показатель округлен до 2 знаков после запятой.

```python
top_cost_per_min = cls_links.top_cost_per_minute(20)
print(top_cost_per_min)
%timeit cls_links.top_cost_per_minute(20)
```

## Результаты методов класса Ratings

```python
filepath = '../datasets/ratings.csv'
cls_ratings_movies = MA.Ratings.Movies(filepath)
%timeit MA.Ratings.Movies(filepath)
```

### Метод dist_by_year() возвращает словарь, где ключи — годы, а значения — количества. Отсортирован по возрастанию лет. Годы извлечены из временных меток.

```python
res = cls_ratings_movies.dist_by_year()
print(res)
%timeit cls_ratings_movies.dist_by_year()
```

### Метод dist_by_rating() возвращает словарь, где ключи — рейтинги, а значения — счетчики. Отсортирован по возрастанию рейтинга.

```python
res = cls_ratings_movies.dist_by_rating()
print(res)
%timeit cls_ratings_movies.dist_by_rating()
```

### Метод top_by_num_of_ratings(n) возвращает top-n фильмов по количеству оценок. Это словарь, где ключами являются названия фильмов, а значениями — числа. Отсортирован по числам в порядке убывания.

```python
res = cls_ratings_movies.top_by_num_of_ratings(20)
print(res)
%timeit cls_ratings_movies.top_by_num_of_ratings(20)
```

### Метод top_by_ratings(n,metric) возвращает top-n фильмов по среднему (metric='average') или медианному (metric='median') значению рейтингов. Это словарь, где ключами являются названия фильмов, а значениями — значения метрик. Отсортирован по убыванию метрик. Значения округлены до 2 десятичных знаков.

```python
res = cls_ratings_movies.top_by_ratings(20,'average')
print(res)
%timeit cls_ratings_movies.top_by_ratings(20,'average')
```

```python
res = cls_ratings_movies.top_by_ratings(20,'median')
print(res)
%timeit cls_ratings_movies.top_by_ratings(20,'median')
```

### Метод top_controversial(n) возвращает top-n фильмов по дисперсии рейтингов. Это словарь, где ключами являются названия фильмов, а значениями — дисперсии. Отсортирован по дисперсии по убыванию. Значения округлены до 2 десятичных знаков.

```python
filepath = '../datasets/ratings.csv'
cls_ratings_users = MA.Ratings.Users(filepath)
%timeit MA.Ratings.Users(filepath)
```

### Метод user_activity() возвращает распределение пользователей по количеству оценок, сделанных ими.

```python
res = cls_ratings_users.user_activity()
print(res)
%timeit cls_ratings_users.user_activity()
```

### Метод user_tendency(metric) возвращает распределение пользователей по средним (metric = 'average') или медианным (metric = 'median') оценкам, сделанным ими.

```python
res = cls_ratings_users.user_tendency('average')
print(res)
%timeit cls_ratings_users.user_tendency('average')
```

```python
res = cls_ratings_users.user_tendency('median')
print(res)
%timeit cls_ratings_users.user_tendency('median')
```

### Метод user_controversial(n) возвращает n лучших пользователей с самой большой дисперсией их оценок.

```python
res = cls_ratings_users.user_controversial(20)
print(res)
%timeit cls_ratings_users.user_controversial(20)
```
